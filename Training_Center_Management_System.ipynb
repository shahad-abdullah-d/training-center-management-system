# Part 2 — Custom Exceptions

Create custom exceptions for:

- duplicate trainee registration,
- course capacity being full,
- trainee not found,
- course not found,
- invalid score.

These exceptions should inherit from `Exception` or `ValueError`.

````python
class DuplicateTraineeError(Exception):
    def __init__(self, trainee_id: int):
        self.trainee_id = trainee_id
        super().__init__(
            f"Trainee with ID {trainee_id} is already registered."
        )


class CourseFullError(Exception):
    def __init__(self, course_id: int):
        self.course_id = course_id
        super().__init__(
            f"Course with ID {course_id} is full."
        )


class TraineeNotFoundError(Exception):
    def __init__(self, trainee_id: int):
        self.trainee_id = trainee_id
        super().__init__(
            f"Trainee with ID {trainee_id} was not found."
        )


class CourseNotFoundError(Exception):
    def __init__(self, course_id: int):
        self.course_id = course_id
        super().__init__(
            f"Course with ID {course_id} was not found."
        )


class InvalidScoreError(ValueError):
    def __init__(self, score: float):
        self.score = score
        super().__init__(
            f"Invalid score: {score}. Score must be between "
            f"{MIN_SCORE} and {MAX_SCORE}."
        )
````

# Part 3 — Validation Functions

Create reusable functions for:

1. cleaning a full name,
2. validating an email address,
3. validating and converting a score,
4. validating a non-empty course code.

## Required Behaviors

### `clean_name(name)`
- remove surrounding spaces,
- replace repeated internal spaces,
- convert to title case.

### `validate_email(email)`
- must contain exactly one `@`,
- must contain a dot after `@`,
- must not contain spaces.

### `validate_score(value)`
- convert the value to `float`,
- accept only values from 0 to 100,
- raise `InvalidScoreError` otherwise.

### `validate_course_code(code)`
- remove spaces,
- convert to uppercase,
- reject an empty value.

````python
def clean_name(name: str) -> str:
    name = name.strip()
    name = " ".join(name.split())
    name = name.title()
    return name

def validate_email(email: str) -> str:
    count = 0

    for char in email:
        if char == " ":
            raise ValueError("Email is not eligible.")

        elif char == "@":
            count += 1

            if count >= 2:
                raise ValueError("Email is not eligible.")

    if count != 1:
        raise ValueError("Email is not eligible.")

    username, domain = email.split("@")

    if "." not in domain:
        raise ValueError("Email is not eligible.")

    return email


def validate_score(value: Any) -> float:

    value = float(value)

    if MIN_SCORE <= value <= MAX_SCORE:
        return value

    raise InvalidScoreError(value)


def validate_course_code(code: str) -> str:

    code = code.strip()

    if len(code) == 0:
        raise ValueError("Course code cannot be empty.")

    code = code.upper()

    return code
````

# Part 5 — Class `Course`

Create a class named `Course`.

## Required Attributes

- `code`
- `title`
- `capacity`
- `passing_score`
- `minimum_attendance`
- `schedule`
- `enrolled_trainees`

Use a tuple for `schedule` and a set for `enrolled_trainees`.

## Required Methods

- `has_space()`
- `enroll_trainee(trainee_id)`
- `is_enrolled(trainee_id)`
- `to_dict()`
- `from_dict()`
- `__str__()`

````python
class Course:
    def __init__(
        self,
        code: str,
        title: str,
        capacity: int,
        passing_score: float = DEFAULT_PASSING_SCORE,
        minimum_attendance: float = DEFAULT_MIN_ATTENDANCE,
        schedule: tuple[str, ...] = (),
    ):
        # TODO

        # نخزن كود الكورس بعد التحقق منه
        self.code = validate_course_code(code)

        # نخزن اسم الكورس
        self.title = title

        # نخزن السعة القصوى للكورس
        self.capacity = capacity

        # نخزن درجة النجاح
        self.passing_score = passing_score

        # نخزن أقل نسبة حضور مطلوبة
        self.minimum_attendance = minimum_attendance

        # نخزن جدول الكورس كـ Tuple لأنه ثابت
        self.schedule = schedule

        # ننشئ Set فاضي لأرقام المتدربين المسجلين
        self.enrolled_trainees = set()


    def has_space(self) -> bool:
        # TODO

        # نرجع True إذا كان عدد المتدربين أقل من سعة الكورس
        return len(self.enrolled_trainees) < self.capacity


    def enroll_trainee(self, trainee_id: int) -> None:
        # TODO: Raise CourseFullError when appropriate.

        # إذا ما فيه مكان داخل الكورس، نرفع خطأ
        if not self.has_space():
            raise CourseFullError(self.code)

        # إذا فيه مكان، نضيف رقم المتدرب داخل الـSet
        self.enrolled_trainees.add(trainee_id)


    def is_enrolled(self, trainee_id: int) -> bool:
        # TODO

        # نتحقق هل رقم المتدرب موجود داخل المسجلين
        return trainee_id in self.enrolled_trainees


    def to_dict(self) -> dict:
        # TODO

        # نحول بيانات الـobject إلى Dictionary حتى يمكن حفظها في JSON
        return {
            "code": self.code,
            "title": self.title,
            "capacity": self.capacity,
            "passing_score": self.passing_score,
            "minimum_attendance": self.minimum_attendance,

            # JSON لا يدعم Tuple، لذلك نحوله إلى List
            "schedule": list(self.schedule),

            # JSON لا يدعم Set، لذلك نحوله إلى List
            "enrolled_trainees": list(self.enrolled_trainees),
        }


    @classmethod
    def from_dict(cls, data: dict) -> "Course":
        # TODO

        # ننشئ Course object جديد من البيانات الموجودة في الـDictionary
        course = cls(
            data["code"],
            data["title"],
            data["capacity"],
            data["passing_score"],
            data["minimum_attendance"],

            # نرجع جدول الكورس من List إلى Tuple
            tuple(data["schedule"]),
        )

        # نرجع المتدربين المسجلين من List إلى Set
        course.enrolled_trainees = set(data["enrolled_trainees"])

        # نرجع الـobject بعد استعادة جميع بياناته
        return course


    def __str__(self) -> str:
        # TODO

        # نحدد شكل بيانات الكورس عند استخدام print()
        return (
            f"Course: {self.code}, "
            f"Title: {self.title}, "
            f"Capacity: {self.capacity}"
        )
````

# Part 6 — Class `TrainingCenter`

This class controls the complete system.

## Required Attributes

- `name`
- `trainees`: dictionary mapping trainee ID to `Trainee`
- `courses`: dictionary mapping course code to `Course`

## Required Methods

### Trainee and course management
- `add_course(course)`
- `register_trainee(trainee)`
- `get_course(course_code)`
- `get_trainee(trainee_id)`
- `enroll_trainee(trainee_id, course_code)`

### Data recording
- `record_attendance(trainee_id, course_code, session_number)`
- `record_score(trainee_id, course_code, score)`

### Calculations
- `attendance_percentage(trainee_id, course_code)`
- `course_status(trainee_id, course_code)`
- `certificate_eligible(trainee_id, course_code)`

### Reports
- `trainee_report(trainee_id)`
- `course_report(course_code)`
- `overall_summary()`

### Files
- `save_json(path)`
- `load_json(path)`
- `export_course_report_csv(course_code, path)`

````python
class TrainingCenter:
    def __init__(self, name: str):
        self.name = name

        # Dictionary: رقم المتدرب هو المفتاح والـTrainee object هو القيمة
        self.trainees: dict[int, Trainee] = {}

        # Dictionary: كود الكورس هو المفتاح والـCourse object هو القيمة
        self.courses: dict[str, Course] = {}


    def add_course(self, course: Course) -> None:
        # TODO

        # نمنع إضافة نفس كود الكورس مرتين
        if course.code in self.courses:
            raise ValueError(f"Course {course.code} already exists.")

        # نخزن الكورس باستخدام الكود كمفتاح
        self.courses[course.code] = course


    def register_trainee(self, trainee: Trainee) -> None:
        # TODO: Prevent duplicate trainee IDs.

        # نتحقق هل رقم المتدرب مسجل مسبقًا
        if trainee.trainee_id in self.trainees:
            raise DuplicateTraineeError(trainee.trainee_id)

        # نخزن المتدرب باستخدام رقمه كمفتاح
        self.trainees[trainee.trainee_id] = trainee


    def get_course(self, course_code: str) -> Course:
        # TODO: Raise CourseNotFoundError when needed.

        # ننظف كود الكورس ونحوله إلى Uppercase
        course_code = validate_course_code(course_code)

        # إذا الكورس غير موجود نرفع خطأ
        if course_code not in self.courses:
            raise CourseNotFoundError(course_code)

        # نرجع Course object
        return self.courses[course_code]


    def get_trainee(self, trainee_id: int) -> Trainee:
        # TODO: Raise TraineeNotFoundError when needed.

        # إذا المتدرب غير موجود نرفع خطأ
        if trainee_id not in self.trainees:
            raise TraineeNotFoundError(trainee_id)

        # نرجع Trainee object
        return self.trainees[trainee_id]


    def enroll_trainee(
        self,
        trainee_id: int,
        course_code: str,
    ) -> None:
        # TODO: Update both Trainee and Course objects.

        # نحصل على المتدرب، وإذا غير موجود تشتغل الـException
        trainee = self.get_trainee(trainee_id)

        # نحصل على الكورس، وإذا غير موجود تشتغل الـException
        course = self.get_course(course_code)

        # نسجل رقم المتدرب داخل Course object
        course.enroll_trainee(trainee_id)

        # نسجل كود الكورس داخل Trainee object
        trainee.enroll(course.code)


    def record_attendance(
        self,
        trainee_id: int,
        course_code: str,
        session_number: int,
    ) -> None:
        # TODO

        # نحصل على المتدرب والكورس
        trainee = self.get_trainee(trainee_id)
        course = self.get_course(course_code)

        # نتأكد أن المتدرب مسجل في الكورس
        if not course.is_enrolled(trainee_id):
            raise ValueError(
                f"Trainee {trainee_id} is not enrolled in {course.code}."
            )

        # نتأكد أن رقم الجلسة موجب
        if session_number <= 0:
            raise ValueError("Session number must be positive.")

        # نسجل الحضور داخل Trainee object
        trainee.record_attendance(course.code, session_number)


    def record_score(
        self,
        trainee_id: int,
        course_code: str,
        score: float,
    ) -> None:
        # TODO

        # نحصل على المتدرب والكورس
        trainee = self.get_trainee(trainee_id)
        course = self.get_course(course_code)

        # نتأكد أن المتدرب مسجل في الكورس
        if not course.is_enrolled(trainee_id):
            raise ValueError(
                f"Trainee {trainee_id} is not enrolled in {course.code}."
            )

        # نضيف الدرجة، وداخل add_score يتم التحقق من صحتها
        trainee.add_score(course.code, score)


    def attendance_percentage(
        self,
        trainee_id: int,
        course_code: str,
    ) -> float:
        # TODO

        # نحصل على المتدرب والكورس
        trainee = self.get_trainee(trainee_id)
        course = self.get_course(course_code)

        # عدد جلسات الكورس الكلي مأخوذ من الـschedule
        total_sessions = len(course.schedule)

        # نتجنب القسمة على صفر إذا لم يوجد جدول
        if total_sessions == 0:
            return 0.0

        # نحصل على جلسات الحضور، أو Set فاضي إذا ما عنده حضور
        attended_sessions = trainee.attendance.get(course.code, set())

        # نحسب النسبة: عدد الجلسات المحضورة ÷ إجمالي الجلسات
        # نحسب النسبة المئوية: (الحضور ÷ إجمالي الجلسات) × 100
        return len(attended_sessions) / total_sessions * 100


    def course_status(
        self,
        trainee_id: int,
        course_code: str,
    ) -> str:
        # TODO: Return Passed, Failed, or Incomplete.

        # نحصل على المتدرب والكورس
        trainee = self.get_trainee(trainee_id)
        course = self.get_course(course_code)

        # نحصل على درجات المتدرب في هذا الكورس
        scores = trainee.scores.get(course.code, [])

        # إذا ما عنده درجات، يعتبر الكورس غير مكتمل
        if len(scores) == 0:
            return "Incomplete"

        # نحسب متوسط الدرجات ونسبة الحضور
        average = trainee.average_score(course.code)
        attendance = self.attendance_percentage(
            trainee_id,
            course.code,
        )

        # النجاح يحتاج الدرجة والحضور معًا
        if (
            average >= course.passing_score
            and attendance >= course.minimum_attendance
        ):
            return "Passed"

        # إذا لم تتحقق شروط النجاح
        return "Failed"


    def certificate_eligible(
        self,
        trainee_id: int,
        course_code: str,
    ) -> bool:
        # TODO

        # يستحق الشهادة فقط إذا كانت حالته Passed
        return self.course_status(trainee_id, course_code) == "Passed"


    def trainee_report(self, trainee_id: int) -> dict:
        # TODO

        # نحصل على المتدرب
        trainee = self.get_trainee(trainee_id)

        # قائمة نخزن فيها تفاصيل كل كورس
        courses_report = []

        # نمر على كل الكورسات المسجل فيها المتدرب
        for course_code in trainee.enrolled_courses:
            course = self.get_course(course_code)

            courses_report.append(
                {
                    "course_code": course.code,
                    "course_title": course.title,
                    "average_score": trainee.average_score(course.code),
                    "attendance_percentage": self.attendance_percentage(
                        trainee_id,
                        course.code,
                    ),
                    "status": self.course_status(
                        trainee_id,
                        course.code,
                    ),
                    "certificate_eligible": self.certificate_eligible(
                        trainee_id,
                        course.code,
                    ),
                }
            )

        # نرجع تقرير المتدرب كاملًا
        return {
            "trainee_id": trainee.trainee_id,
            "name": trainee.name,
            "email": trainee.email,
            "courses": courses_report,
        }


    def course_report(self, course_code: str) -> list[dict]:
        # TODO

        # نحصل على الكورس
        course = self.get_course(course_code)

        # قائمة تقرير جميع المتدربين في الكورس
        report = []

        # نمر على أرقام المتدربين المسجلين
        for trainee_id in course.enrolled_trainees:
            trainee = self.get_trainee(trainee_id)

            report.append(
                {
                    "trainee_id": trainee.trainee_id,
                    "name": trainee.name,
                    "average_score": trainee.average_score(course.code),
                    "attendance_percentage": self.attendance_percentage(
                        trainee_id,
                        course.code,
                    ),
                    "status": self.course_status(
                        trainee_id,
                        course.code,
                    ),
                    "certificate_eligible": self.certificate_eligible(
                        trainee_id,
                        course.code,
                    ),
                }
            )

        # نرجع قائمة التقارير
        return report


    def overall_summary(self) -> dict:
        # TODO

        # نحسب إجمالي التسجيلات في جميع الكورسات
        total_enrollments = sum(
            len(course.enrolled_trainees)
            for course in self.courses.values()
        )

        # نرجع ملخص المركز
        return {
            "center_name": self.name,
            "total_trainees": len(self.trainees),
            "total_courses": len(self.courses),
            "total_enrollments": total_enrollments,
        }


    def save_json(self, path: Path) -> None:
        # TODO

        # نحول جميع بيانات النظام إلى بيانات قابلة للحفظ
        data = {
            "name": self.name,
            "trainees": [
                trainee.to_dict()
                for trainee in self.trainees.values()
            ],
            "courses": [
                course.to_dict()
                for course in self.courses.values()
            ],
        }

        # نفتح الملف بوضع الكتابة ونحفظ البيانات
        with open(path, "w", encoding="utf-8") as file:
            json.dump(
                data,
                file,
                indent=4,
                ensure_ascii=False,
            )


    @classmethod
    def load_json(cls, path: Path) -> "TrainingCenter":
        # TODO

        # نفتح ملف JSON ونقرأ البيانات
        with open(path, "r", encoding="utf-8") as file:
            data = json.load(file)

        # ننشئ TrainingCenter جديد
        center = cls(data["name"])

        # نرجع بيانات المتدربين إلى Trainee objects
        for trainee_data in data["trainees"]:
            trainee = Trainee.from_dict(trainee_data)
            center.trainees[trainee.trainee_id] = trainee

        # نرجع بيانات الكورسات إلى Course objects
        for course_data in data["courses"]:
            course = Course.from_dict(course_data)
            center.courses[course.code] = course

        # نرجع المركز بعد استعادة بياناته
        return center


    def export_course_report_csv(
        self,
        course_code: str,
        path: Path,
    ) -> None:
        # TODO

        # نحصل على تقرير الكورس
        report = self.course_report(course_code)

        # أسماء الأعمدة داخل ملف CSV
        fieldnames = [
            "trainee_id",
            "name",
            "average_score",
            "attendance_percentage",
            "status",
            "certificate_eligible",
        ]

        # نفتح ملف CSV بوضع الكتابة
        with open(
            path,
            "w",
            newline="",
            encoding="utf-8",
        ) as file:
            writer = csv.DictWriter(
                file,
                fieldnames=fieldnames,
            )

            # نكتب أسماء الأعمدة
            writer.writeheader()

            # نكتب بيانات المتدربين
            writer.writerows(report)
````

# Part 16 — Overall Management Summary

Create a summary containing:

- number of courses,
- number of registered trainees,
- total enrollments,
- number of passed trainee-course records,
- number of failed trainee-course records,
- number of incomplete trainee-course records,
- course with the highest average,
- total number of certificates eligible.

````python
# TODO: Generate and print the overall summary.

total_enrollments = 0
passed_count = 0
failed_count = 0
incomplete_count = 0
certificate_count = 0

highest_course = None
highest_course_average = -1


# نمر على جميع الكورسات
for course in center.courses.values():

    # نضيف عدد المسجلين في الكورس إلى إجمالي التسجيلات
    total_enrollments += len(course.enrolled_trainees)

    course_averages = []

    # نمر على المتدربين المسجلين في الكورس
    for trainee_id in course.enrolled_trainees:

        trainee = center.get_trainee(trainee_id)

        # نحصل على حالة المتدرب في الكورس
        status = center.course_status(trainee_id, course.code)

        if status == "Passed":
            passed_count += 1

        elif status == "Failed":
            failed_count += 1

        else:
            incomplete_count += 1

        # نحسب عدد المستحقين للشهادة
        if center.certificate_eligible(trainee_id, course.code):
            certificate_count += 1

        # نضيف المتوسط إذا كان للمتدرب درجات
        if course.code in trainee.scores:
            course_averages.append(
                trainee.average_score(course.code)
            )

    # نحسب متوسط الكورس كاملًا
    if len(course_averages) > 0:
        course_average = sum(course_averages) / len(course_averages)

        # نحتفظ بالكورس صاحب أعلى متوسط
        if course_average > highest_course_average:
            highest_course_average = course_average
            highest_course = course.code


# ننشئ الملخص النهائي
summary = {
    "number_of_courses": len(center.courses),
    "number_of_trainees": len(center.trainees),
    "total_enrollments": total_enrollments,
    "passed_records": passed_count,
    "failed_records": failed_count,
    "incomplete_records": incomplete_count,
    "highest_average_course": highest_course,
    "highest_course_average": highest_course_average,
    "certificates_eligible": certificate_count,
}


# نطبع الملخص بشكل مرتب
print(json.dumps(summary, indent=2))
````
